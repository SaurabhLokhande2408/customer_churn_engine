```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

target_col = "Churn Score"
if target_col not in df.columns:
    raise KeyError(f"{target_col!r} is not in df. Available columns: {df.columns.tolist()}")

charges_bin = pd.qcut(df[target_col], q=4, labels=False, duplicates="drop")
chi2_results = {}

for col in catergorical_features.columns:
    contingency = pd.crosstab(df[col], charges_bin)
    if contingency.shape[0] < 2 or contingency.shape[1] < 2:
        continue

    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (evidence of association)' if p_val < alpha else 'Fail to Reject Null (insufficient evidence of association)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame.from_dict(chi2_results, orient='index')
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

```python
#now using pearson correlation to see how our features are correlated with each other taken from chatgpt
from scipy.stats import pearsonr

# ----------------------------------
# Pearson Correlation Calculation
# ----------------------------------

# List of features to check against target
selected_features = ['Age', 'Population', 'Number of Referrals',
       'Tenure in Months', 'Avg Monthly Long Distance Charges',
       'Avg Monthly GB Download', 'Monthly Charge', 'Total Charges',
       'Total Refunds', 'Total Extra Data Charges',
       'Total Long Distance Charges',  'Satisfaction Score', 'CLTV'
    
]

correlations = {
    feature: pearsonr(df[feature], df['Churn Score'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```
